```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display(self):
        print(f"{self.name:<20} Rs. {self.price:.2f}")


def main():
    products = [
        Product("Laptop", 55000),
        Product("Mobile", 18000),
        Product("Headphones", 2500),
        Product("Keyboard", 1200),
        Product("Smart Watch", 3500)
    ]

    while True:
        print("\n===== PRODUCT PRICE COMPARATOR =====")
        print("1. Display All Products")
        print("2. Sort Products by Price")
        print("3. Search Product")
        print("4. Find Cheapest Product")
        print("5. Exit")

        try:
            choice = input("Enter your choice: ").strip()
            if not choice:
                continue

            if choice == '1':
                print("\nAvailable Products:")
                for p in products:
                    p.display()

            elif choice == '2':
                # Sort using python's built-in sorting (comparable to Java's Arrays.sort)
                sorted_products = sorted(products, key=lambda x: x.price)
                print("\nProducts Sorted by Price:")
                for p in sorted_products:
                    p.display()

            elif choice == '3':
                search = input("Enter product name to search: ").strip()
                found = False
                for p in products:
                    if p.name.lower() == search.lower():
                        print("Product Found:")
                        p.display()
                        found = True
                        break
                if not found:
                    print("Product Not Found!")

            elif choice == '4':
                cheapest = min(products, key=lambda x: x.price)
                print("\nCheapest Product:")
                cheapest.display()

            elif choice == '5':
                print("Thank you for using the Comparator!")
                break
            else:
                print("Invalid Choice!")
        except (KeyboardInterrupt, EOFError):
            print("\nProgram terminated.")
            break
if __name__ == '__main__':
    main()
```
